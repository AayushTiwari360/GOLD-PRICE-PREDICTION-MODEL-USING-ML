#**CORELATION**
# Correlation in machine learning refers to the statistical relationship between two variables. It helps determine how changes in one variable are associated with changes in another.
Correlation helps identify whether features are positively, negatively, or not related at all.


```python
correlation = gold.select_dtypes(include=[np.number]).corr()
```

# **Heatmap**
#Hepls invisualization using the Seaborn library, showing the correlation values between variables in a datase.

```python
plt.figure(figsize=(6,16))
sns.heatmap(correlation,cbar=True,square=True,annot=True,annot_kws={'size':8},cmap="plasma")
```

# **correlation['GLD']**
In **Python**, if correlation is a DataFrame containing correlation coefficients between different variables, correlation['GLD'] will extract the correlation values of the variable "GLD" with all other variables in the DataFrame.



```python
# correlation values of GLD
print(correlation['GLD'])
```

```python
# checking Price distrbution of gld
sns.distplot(gold['GLD'],color="green")
```

```python
# splitting the features and target
X = gold.drop(['Date','GLD'],axis=1)
Y = gold['GLD']
```

```python
X.shape
```

```python
Y.shape
```

```python
#train model
train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
#train model
X_train, X_test, y_train, y_test = train_test_split(X,Y,test_size=0.2,random_state=2) # Assign the results of the function to the respective variables
reg = RandomForestRegressor()
reg.fit(X_train,y_train)
pred = reg.predict(X_test)
r2_score(y_test,pred)
```
